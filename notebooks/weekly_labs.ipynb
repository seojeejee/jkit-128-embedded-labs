```python
#1027 외부 인터럽트 (External Interrupt)

# 0x0_ -> 여기 숫자에 따라 달라짐(하위비트)
# 0x01(눌렀다가 뗐을때), 0x02(누르는 순간) -> sw1 ISR(INT4_vect)
# 0x04(눌렀다가 뗐을때), 0x08(누르는 순간) -> sw2 ISR(INT5_vect)
# sw1 - EIMSK = 0x10;
# sw2 - EIMSK = 0x20;

#include <avr/io.h>
#include <avr/interrupt.h>
volatile unsigned char counter;
int main()
 {
	 DDRA = 0xFF;
	 PORTA = 0x00;
	 // header file,
	 // interrupt 관련 함수들이 정의되어 있음.
	 // global variable
	 // port A를 출력 port로 설정.
	 // PORTA에 0xFF값을 출력시킴. 모든 LED를 OFF시킴.
	 // --- external interrupt INT4 관련 register 설정 ----------
	 EICRA = 0x00;
	 EICRB = 0x04;
	 // ISC41 = 1, ISC40 = 0 ==> falling edge
	 EIMSK = 0x20;
	 sei();
	 // INT4 = 1
	 // enable global interrupt
	 // ------------------------------------------------
	 counter = 0x00;
	 // counter를 초기화
	 do {
		 PORTA = counter;
	 }
	 while (1);
	 return (1);
	 // counter값을 LED에 출력
 }
 // --- external interrupt INT0에 대한 service routine --
 ISR(INT5_vect)
 {
	 ++counter;
 }
 // counter값을 증가시킴

#[응용 3] Interrupt를 이용하여 스위치 SW1을 누르면 LED에 나타나는 2진수값이 증가하고, SW2를 누르면 감소하는 프로그램을 작성하고 시험하시오
#include <avr/io.h>
#include <avr/interrupt.h>
volatile unsigned char counter;  // global variable

int main()
{
	DDRA = 0xFF;      // port A를 출력 port로 설정
	PORTA = 0x00;     // PORTA에 0xFF값을 출력시킴. 모든 LED를 OFF시킴

	DDRE = 0x00;      // PE4, PE5 입력으로 설정 (전체 입력)
	PORTE = (1 << PE4) | (1 << PE5);   // 내부 풀업 저항 활성화

	// --- external interrupt register 설정 ---------
	EICRB = 0x0A;     // ISC41 = 1, ISC40 = 1 ==> falling edge
	EIMSK = 0x30;     // INT4, INT5 활성화 (0b00110000)
	sei();            // enable global interrupt

	counter = 0x00; // counter를 초기화
	do {
		PORTA = counter;
	   }
	while (1);
	return (1);
}

// --- SW1: INT4 (증가) ---
ISR(INT4_vect)
{
	counter++;
}

// --- SW2: INT5 (감소) ---
ISR(INT5_vect)
{
	counter--;
}

#include <avr/io.h>
#include <avr/interrupt.h>

// 7-세그먼트 표시 숫자를 저장할 전역 변수 (인터럽트에서 접근)
volatile uint8_t counter = 0;

// 공통 캐소드용 7세그먼트 배열
uint8_t seg[10] = {
	0x3F, // 0
	0x06, // 1
	0x5B, // 2
	0x4F, // 3
	0x66, // 4
	0x6D, // 5
	0x7D, // 6
	0x07, // 7
	0x7F, // 8
	0x6F  // 9
};

// SW1 (증가) - INT4 (PE4) 인터럽트 서비스 루틴
ISR(INT4_vect)
{
	// 디바운싱 처리가 없으므로, 버튼을 누를 때 튀는 현상으로 여러 번 증가할 수 있습니다.
	if (counter < 9)
	counter++;
	else
	counter = 0;

	// 7-세그먼트 값 업데이트
	PORTC = seg[counter];
}

// SW2 (감소) - INT5 (PE5) 인터럽트 서비스 루틴
ISR(INT5_vect)
{
	if (counter > 0)
	counter--;
	else
	counter = 9;

	// 7-세그먼트 값 업데이트
	PORTC = seg[counter];
}

#[응용 4] Interrupt를 이용하여 스위치 SW1을 누르면 7-segment LED에 나타나는 10진수값이 증가하고, SW2를 누르면 감소하는 프로그램을 작성하고 시험하시오
int main(void)
{
	// 1. I/O 포트 설정
	DDRC = 0xFF;    // PORTC 전체를 출력으로 설정 (7-세그먼트)
	// PORTG 설정은 그대로 유지 (다른 기능이 없다면 무시 가능)
	DDRG |= 0x0F;
	PORTG = 0x01;

	// 초기값 설정
	PORTC = seg[0]; // 7-세그먼트에 초기값 '0' 표시
	counter = 0;    // 카운터 변수 초기화

	// 2. 외부 인터럽트 포트 설정 (스위치 입력)
	DDRE = 0x00; // PORTE 전체를 입력으로 설정 (PE4, PE5 포함)
	// PE4 (INT4)와 PE5 (INT5)에 내부 풀업 저항 활성화
	// 스위치를 핀과 GND 사이에 연결하는 경우 사용
	PORTE = (1 << PE4) | (1 << PE5);

	// 3. 외부 인터럽트 제어 레지스터 설정
	// EICRB: INT5, INT4의 트리거 방식을 설정
	// 0x2A (0010 1010): INT5 = Falling Edge (하강 에지), INT4 = Falling Edge (하강 에지)
	// '하강 에지'는 버튼을 누를 때 (HIGH -> LOW) 인터럽트를 발생시키는 가장 일반적인 방식입니다.
	EICRB = 0x2A;

	// EIMSK: INT4, INT5 인터럽트 활성화
	EIMSK = (1 << INT4) | (1 << INT5);

	// 4. 전역 인터럽트 활성화
	sei();

	// 메인 루프는 인터럽트 설정을 완료했기 때문에 대기합니다.
	while (1)
	{
		// 모든 스위치 처리는 ISR에서 수행됩니다.
	}

	return 0;
}

#[응용 5] SW1을 누를 때마다 7-segment LED에 표시되는 10진수가 1씩 증가하고 SW2를 누르면 그 순간부터 1초마다 1씩 감소하고 0이되는 순간 감소가 중단되고 buzzer가 1초씩 세 번 울리는 타이머기능의 프로그램을 작성하시오. 여기서, SW1는 타이머의 시간설정기능이고 SW2는 타이머의 동작을 시작하게 하는 기능을 수행한다. 실험키트에 전원을 공급한 직후에는 7-segment LED에 0이 표시되도록 한다. 단, switch의 입력을 external interrupt 방식을 사용한다

#define F_CPU 16000000UL
#include <avr/io.h>
#include <avr/interrupt.h>
#include <util/delay.h>

// =======================================================
// 핀 정의 및 설정
// =======================================================
#define SW1_PIN PE4 // INT4
#define SW2_PIN PE5 // INT5
#define SEG_PORT PORTC
#define SEG_DDR  DDRC
#define DIGIT_PORT PORTG
#define BUZZER_PIN PB4
#define BUZZER_PORT PORTB
#define BUZZER_DDR DDRB

// =======================================================
// 전역 변수 및 7-Segment 코드
// =======================================================
volatile uint8_t timer_value = 0;
volatile uint8_t timer_running = 0;
volatile uint8_t buzzer_countdown = 0; // 3회 울림 카운터

const uint8_t segment_codes[] = {
	0x3F, 0x06, 0x5B, 0x4F, 0x66,
	0x6D, 0x7D, 0x07, 0x7F, 0x6F
};

// =======================================================
// 부저 제어 함수 (TIMER3를 ON/OFF)
// =======================================================
void control_buzzer(uint8_t state) {
	if (state) {
		TCCR3B |= (1 << CS32); // 분주비 256 설정 (TIMER3 시작)
		} else {
		TCCR3B &= ~((1 << CS32) | (1 << CS31) | (1 << CS30)); // TIMER3 정지
		BUZZER_PORT &= ~(1 << BUZZER_PIN); // PB4 LOW로 고정
	}
}

void display_7_segment(uint8_t number) {
	if (number <= 9) {
		SEG_PORT = segment_codes[number];
	}
}

// SW1/SW2 ISR은 이전과 동일
ISR(INT4_vect) { _delay_ms(50); if (!timer_running) { timer_value = (timer_value + 1) % 10; display_7_segment(timer_value); } while (!(PINE & (1 << SW1_PIN))); }
ISR(INT5_vect) { _delay_ms(50); if (timer_value > 0 && !timer_running) { timer_running = 1; } while (!(PINE & (1 << SW2_PIN))); }

// =======================================================
// 타이머 3 인터럽트 서비스 루틴 (부저 주파수 발생)
// =======================================================
ISR(TIMER3_COMPA_vect) {
	BUZZER_PORT ^= (1 << BUZZER_PIN);
}

// =======================================================
// 타이머 1 인터럽트 서비스 루틴 (1초 주기 제어 - 핵심 수정)
// =======================================================
ISR(TIMER1_COMPA_vect) {

	// 1. 타이머 감소 로직
	if (timer_running) {
		if (timer_value > 0) {
			timer_value--;
			display_7_segment(timer_value);
			} else { // timer_value == 0: 감소 중단 및 부저 초기화
			timer_running = 0;
			buzzer_countdown = 3; // 3회 울림 설정
			control_buzzer(1);    // 부저 즉시 켜기 (첫 번째 1초 ON 시작)
		}
	}

	// 2. 부저 1초 간격 제어 로직 (buzzer_countdown이 0보다 클 때만 실행)
	else if (buzzer_countdown > 0 || (BUZZER_PORT & (1 << BUZZER_PIN))) {

		// 현재 부저 ON 상태인지 확인 (PB4 핀 상태 확인)
		if (BUZZER_PORT & (1 << BUZZER_PIN)) {
			// 부저가 켜져 있으면 (1초 ON 끝) -> 꺼야 할 차례
			control_buzzer(0); // 부저 끄기 (1초 OFF 시작)
			buzzer_countdown--; // 🚨 ON이 끝나면 카운트 감소
			} else {
			// 부저가 꺼져 있으면 (1초 OFF 끝) -> 다시 켜야 할 차례
			// 단, 카운트가 0이 아니어야 다음 ON을 시도합니다.
			if (buzzer_countdown > 0) {
				control_buzzer(1); // 부저 켜기 (다음 1초 ON 시작)
				} else {
				// 마지막 OFF 상태이고 카운트가 0이면, 다시 켤 필요 없이 종료
				control_buzzer(0);
			}
		}
	}

	// 3. 최종 종료
	else {
		control_buzzer(0); // 완전히 종료
	}
}

// =======================================================
// Main 함수
// =======================================================
int main(void) {
	// 1. I/O 포트 설정
	SEG_DDR = 0xFF;
	DDRG |= 0x01;
	BUZZER_DDR |= (1 << BUZZER_PIN);

	DDRE &= ~((1 << SW1_PIN) | (1 << SW2_PIN));
	PORTE |= ((1 << SW1_PIN) | (1 << SW2_PIN));

	// 2. 초기 상태 설정
	PORTG = 0x01;
	timer_value = 0;
	display_7_segment(timer_value);
	control_buzzer(0);

	// 3. 외부 인터럽트 설정 (INT4, INT5)
	EICRB = (EICRB & ~((1 << ISC51) | (1 << ISC50) | (1 << ISC41) | (1 << ISC40))) |
	(1 << ISC51) | (0 << ISC50) |
	(1 << ISC41) | (0 << ISC40);
	EIMSK |= (1 << INT5) | (1 << INT4);

	// 4. 타이머/카운터 1 설정 (1초 주기 제어)
	TCCR1B = 0;
	TCCR1B |= (1 << WGM12);
	OCR1A = 15624;
	TIMSK |= (1 << OCIE1A); //
	TCCR1B |= (1 << CS12) | (1 << CS10);

	// 5. 타이머/카운터 3 설정 (부저 주파수 발생)
	TCCR3B = 0;
	TCCR3B |= (1 << WGM32);
	OCR3A = 31;
	ETIMSK |= (1 << OCIE3A); //
	// TCCR3B 분주비는 control_buzzer 함수에서 설정됨

	// 6. 전체 인터럽트 활성화
	sei();

	do {
	} while (1);

	return 0;
}
```

```python
#20251103 [응용 1] 0.5초 간격으로 LED를 LD1 -> --- --> LD8 ->LD1 의 순으로 Rotate Right 시키면서 이동 점등시키는 프로그램을 작성하시오.
#define F_CPU 16000000UL   // 16MHz 클럭
#include <avr/io.h>
#include <util/delay.h>

int main(void)
{
	unsigned char led = 0x01;  // 초기 LD1 ON
	DDRA = 0xFF;               // PORTA를 출력으로 설정

	while (1)
	{
		PORTA = led;           // LED 점등
		_delay_ms(500);        // 0.5초 대기

		// Rotate Right
		if (led & 0x01)        // 최하위 비트가 1이면 (LD8에서 다시 LD1로)
		led = 0x80;        // LD8 → LD1로 회전
		else
		led >>= 1;         // 오른쪽으로 한 칸씩 이동
	}
}

#[응용 2] 동일한 프로그램에 대해 타이머의 prescaler값을 다른 값으로 변경하여 위의 실험을 반복하고 결과를 관찰하고 분석하시오. TCCR0 레지스터에서 prescaler를 설정할 수 있다.
TCCR0 = (1 << CS00); #매우빠름
TCCR0 = (1 << CS01); #빠름
TCCR0 = (1 << CS01) | (1 << CS00); #중간속도
TCCR0 = (1 << CS02); #느림
TCCR0 = (1 << CS02) | (1 << CS00); #매우느림

#[응용 3] SW1을 누르면 1초 간격으로, SW2를 누르면 0.5초 간격으로 모든 LED를 점등시키는 프로그램을 작성하시오.
#define F_CPU 16000000UL
#include <avr/io.h>
#include <avr/interrupt.h>

#define SW1_PIN PE4
#define SW2_PIN PE5

volatile uint8_t led_state = 0xFF;     // LED 상태 (ON/OFF)
volatile uint16_t overflow_count = 0;  // 오버플로 카운터
volatile uint16_t target_count = 61;   // 기본 1초 (SW1)

ISR(TIMER0_OVF_vect)
{
	overflow_count++;

	if (overflow_count >= target_count)
	{
		PORTA = led_state;     // LED 출력
		led_state = ~led_state; // LED 상태 반전
		overflow_count = 0;    // 카운터 초기화
	}
}

int main(void)
{
	DDRA = 0xFF;   // LED 출력 설정
	DDRE &= ~((1 << SW1_PIN) | (1 << SW2_PIN));  // 스위치 입력 설정
	PORTE |= (1 << SW1_PIN) | (1 << SW2_PIN);    // 내부 풀업 활성화

	// -----------------------------
	// Timer0 설정 (Normal mode)
	// -----------------------------
	TCCR0 = (1 << CS02) | (1 << CS00); // 분주비 1024
	TIMSK |= (1 << TOIE0);             // Overflow interrupt enable
	TCNT0 = 0;                         // 초기화

	sei(); // 전역 인터럽트 허용

	while (1)
	{
		// 스위치 입력 확인 (Active Low)
		if (!(PINE & (1 << SW1_PIN)))
		target_count = 61;   // 1초
		else if (!(PINE & (1 << SW2_PIN)))
		target_count = 30;   // 0.5초
	}
}

#[응용 4] 7 segment LED에 왼쪽 두 개의 LED에는 분, 오른쪽 두 개의 LED에는 초의 시간을 출력하는 프로그램을 작성하시오
#include <avr/io.h>

int main(void)
{
	volatile unsigned long int delay;
	unsigned char m1, m2, s1, s2;
	int min = 0, sec = 0;

	// 7-segment 숫자 패턴 (공통 캐소드 기준)
	uint8_t seg[10] = {
		0x3F, // 0
		0x06, // 1
		0x5B, // 2
		0x4F, // 3
		0x66, // 4
		0x6D, // 5
		0x7D, // 6
		0x07, // 7
		0x7F, // 8
		0x6F  // 9
	};

	DDRC = 0xFF;  // 세그먼트 데이터 출력 (a~g)
	DDRG |= 0x0F; // 4자리 선택 출력 (PG0~PG3)

	while (1) {
		// 분/초 → 4자리 분리
		m1 = min / 10;
		m2 = min % 10;
		s1 = sec / 10;
		s2 = sec % 10;

		// 1초 동안 4자리 멀티플렉싱 반복
		for (int repeat = 0; repeat < 200; repeat++) {
			// 1의 자리(초)
			PORTG = 0x01;       // 자리 선택
			PORTC = seg[s2];    // 숫자 출력
			for (delay = 0; delay < 500; delay++); // 짧은 지연

			// 10의 자리(초)
			PORTG = 0x02;
			PORTC = seg[s1];
			for (delay = 0; delay < 500; delay++);

			// 1의 자리(분)
			PORTG = 0x04;
			PORTC = seg[m2];
			for (delay = 0; delay < 500; delay++);

			// 10의 자리(분)
			PORTG = 0x08;
			PORTC = seg[m1];
			for (delay = 0; delay < 500; delay++);
		}

		// 1초 증가
		sec++;
		if (sec >= 60) {
			sec = 0;
			min++;
			if (min >= 60) min = 0;
		}
	}
}

#[응용 5] ATmega128 datasheet의 16-bit timer/counter 부분을 참조하여 Timer1을 이용하여 [응용 1], [응용 3], [응용 4]를 반복하시오.
#8비트 → 1회 오버플로우만으로 1초 이상을 구현하기 어려움 (0 ~ 255, TCNT0, TCNT2)
→ 1초 이상 필요하면 반복 계산 또는 분주 비율 조정 필요
#16비트 → 1초 단위도 한 번의 Compare Match 또는 오버플로우로 가능 (0 ~ 65535, TCNT1)
→ 계산 간단, 정확도 높음“8비트 = 짧은 간격, 반복 필요 / 16비트 = 긴 간격, 정확하고 반복 적게 사용 가능”

#[응용 6] Buzzer에서 8음계 (도, 레, 미, 파, 솔, 라, 시, 도)의 음이 순서대로 2초씩 들리는 프로그램을 작성하시오. 각 음계에 대한 주파수 및 주기를 조사한 후 각 buzzer를 구동하기 위해 Timer0를 이용하여 시간을 지연시키는 기능을 구현한다. 여기에서 for loop나 delay 관련 library 함수는 사용하지 않는다.
#define F_CPU 16000000UL
#include <avr/io.h>
#include <avr/interrupt.h>
#include <stdint.h>

#define BUZZER PB4 // PB4 핀 사용

// 8음계 주파수 (Hz)
const uint16_t freq[8] = {261, 294, 329, 349, 392, 440, 493, 523};

volatile uint8_t note_index = 0;
volatile uint16_t tick_counter = 0; // Timer0 누적 카운트

// ----------------------------------------------------------------------
// Timer1 Compare Match ISR: Buzzer Square Wave 생성 (주파수)
// ----------------------------------------------------------------------
ISR(TIMER1_COMPA_vect)
{
	PORTB ^= (1 << BUZZER); // PB4 토글
}

// ----------------------------------------------------------------------
// Timer0 Compare Match ISR: 16초 카운트 및 음계 전환 (수정됨)
// ----------------------------------------------------------------------
ISR(TIMER0_COMP_vect)
{
	tick_counter++;

	// 16초 경과 시 다음 음계로 전환
	// OCR0 = 156 → ISR 호출 약 0.01초 → 16초 = 1600 ticks
	if(tick_counter >= 1600) // 💡 100 -> 1600으로 수정
	{
		tick_counter = 0; // 초기화
		note_index++;
		if(note_index >= 8)
		note_index = 0;

		// Timer1 OCR1A 재설정: 주파수에 맞게
		OCR1A = (uint16_t)((F_CPU / (2UL * 64UL * freq[note_index])) - 1);
	}
}

int main(void)
{
	// 포트 설정
	DDRB |= (1 << BUZZER); // PB4 출력
	PORTB &= ~(1 << BUZZER);

	// ----------------------------------------------------------------------
	// Timer1 설정: 16비트, CTC, 분주 64 → Buzzer 주파수
	// ----------------------------------------------------------------------
	TCCR1B = (1 << WGM12) | (1 << CS11) | (1 << CS10); // CTC, prescaler 64
	OCR1A = (uint16_t)((F_CPU / (2UL * 64UL * freq[note_index])) - 1);
	TIMSK |= (1 << OCIE1A); // Timer1 Compare Match A 인터럽트 활성화

	// ----------------------------------------------------------------------
	// Timer0 설정: 8비트, CTC, 분주 1024 → 0.01초 단위 ISR
	// ----------------------------------------------------------------------
	TCCR0 = (1 << WGM01) | (1 << CS02) | (1 << CS00); // CTC, prescaler 1024
	OCR0 = 156; // 약 0.01초 per ISR
	TIMSK |= (1 << OCIE0); // Timer0 Compare Match 인터럽트 활성화

	sei(); // 전역 인터럽트 허용

	while(1)
	{
		// 메인 루프 비움, ISR에서 처리
	}
}
```

```python
#20251110 [응용 1] 인터럽트 방식을 이용하여 0.5초 간격으로 LED를 LD1 -> LD8 ->LD1의 순으로 Rotate Right 시키면서 이동 점등시키는 프로그램을 작성하시오.

#include <avr/io.h>
#include <avr/interrupt.h>
#define OVERFLOW 256
#define TICKS_PER_SEC 500   // 0.5초 간격 (기존 1000에서 절반으로 변경)

volatile unsigned int tick_time = 0;
volatile unsigned char led = 0x01;   // LD1부터 시작

ISR(TIMER0_OVF_vect)
{
	TCNT0 = OVERFLOW - 125;  // 타이머 카운터 초기값 (시간 조정용)
	tick_time++;
	if (tick_time >= TICKS_PER_SEC) {
		tick_time = 0;

		//LED Rotate Right (LD1 → LD2 → ... → LD8 → LD1)
		led >>= 1;
		if (led == 0x00)
		led = 0x80;  // 끝까지 이동하면 다시 LD8 → LD1로 순환

		PORTA = led;
	}
}

int main(void)
{
	DDRA = 0xFF;      // PORTA를 출력 포트로 설정
	PORTA = 0x01;     // LD1 ON (시작점)
	TCNT0 = OVERFLOW - 125;  // 초기 타이머 카운터 설정

	TIMSK = 0x01;     // 타이머0 오버플로우 인터럽트 허용
	sei();             // 전역 인터럽트 허용
	TCCR0 = 0x05;     // 분주비 1024, Normal 모드

	while (1) {
		// 메인 루프는 비워둠 (인터럽트에 의해 LED가 자동으로 제어됨)
	}

	return 0;
}

#[응용 2] 인터럽트 방식을 이용하여 7 segment LED에 왼쪽 2개의 LED에는 분, 오른쪽 2개의 LED에는 초의 시간을 출력하는 프로그램을 작성하시오.
#define F_CPU 16000000UL
#include <avr/io.h>
#include <avr/interrupt.h>

volatile unsigned char sec = 0;
volatile unsigned char min = 0;
volatile unsigned int tick_time = 0;

#define OVERFLOW 256
#define TICKS_PER_SEC 1000   // 1ms × 1000 = 1초

// 7-segment 숫자 패턴 (공통 캐소드 기준)
uint8_t seg[10] = {
	0x3F, // 0
	0x06, // 1
	0x5B, // 2
	0x4F, // 3
	0x66, // 4
	0x6D, // 5
	0x7D, // 6
	0x07, // 7
	0x7F, // 8
	0x6F  // 9
};

// ----------------------
// Timer0 오버플로우 인터럽트 (약 1ms 주기)
// ----------------------
ISR(TIMER0_OVF_vect)
{
	TCNT0 = OVERFLOW - 125;  // 1ms 보정
	tick_time++;

	if (tick_time >= TICKS_PER_SEC) { // 1초 경과 시
		tick_time = 0;
		sec++;
		if (sec >= 60) {
			sec = 0;
			min++;
			if (min >= 60) min = 0;
		}
	}
}

// ----------------------
// 초기화 함수
// ----------------------
void Timer0_Init(void)
{
	TCNT0 = OVERFLOW - 125;
	TIMSK = 0x01;      // Timer0 Overflow 인터럽트 허용
	TCCR0 = 0x05;      // 분주비 1024, Normal 모드
	sei();             // 전역 인터럽트 허용
}

// ----------------------
// 7세그먼트 표시 함수
// ----------------------
void display_FND(void)
{
	volatile unsigned long int delay;
	unsigned char m1, m2, s1, s2;

	m1 = min / 10;
	m2 = min % 10;
	s1 = sec / 10;
	s2 = sec % 10;

	// 빠르게 4자리 순차 표시 (잔상효과)
	PORTG = 0x01;
	PORTC = seg[s2];
	for (delay = 0; delay < 300; delay++);

	PORTG = 0x02;
	PORTC = seg[s1];
	for (delay = 0; delay < 300; delay++);

	PORTG = 0x04;
	PORTC = seg[m2];
	for (delay = 0; delay < 300; delay++);

	PORTG = 0x08;
	PORTC = seg[m1];
	for (delay = 0; delay < 300; delay++);
}

// ----------------------
// 메인 함수
// ----------------------
int main(void)
{
	DDRC = 0xFF;   // 세그먼트 데이터 출력
	DDRG |= 0x0F;  // 자리 선택 출력

	Timer0_Init();  // 타이머0 설정 및 인터럽트 시작

	while (1) {
		display_FND(); // 실시간 표시 (1초 증가 시 반영)
	}
}

#[응용 4] 7-segment LED에 분과 초를 표시하고 1초 단위로 증가하는 60분 시계 프로그램을 작성하시오. 단, 타이머0 인터럽트를 이용하여 7-segment LED를 동적으로 구동하는데, 숫자가 흔들리지 않고 안정적으로 보이도록 시간을 조절한다. 1초 단위의 시간지연은 타이머1 인터럽트를 사용한다. 이 방법의 특징을 [응용 2]의 프로그램과 비교하여 설명하시오
#define F_CPU 16000000UL
#include <avr/io.h>
#include <avr/interrupt.h>

volatile unsigned char sec = 0;    // 초
volatile unsigned char min = 0;    // 분
volatile unsigned char digit = 0;  // 현재 표시 중인 자리 (0~3)
volatile unsigned char fnd_data[4]; // 표시할 4자리 값 저장용

// 7-segment 숫자 패턴 (공통 캐소드)
uint8_t seg[10] = {
	0x3F, // 0
	0x06, // 1
	0x5B, // 2
	0x4F, // 3
	0x66, // 4
	0x6D, // 5
	0x7D, // 6
	0x07, // 7
	0x7F, // 8
	0x6F  // 9
};

// ===========================================================
// Timer0 인터럽트 — 약 1ms마다 호출 → 7-segment 표시용
// ===========================================================
ISR(TIMER0_OVF_vect)
{
	TCNT0 = 256 - 125; // 약 1ms 간격 (16MHz / 1024 분주 기준)

	// 현재 표시할 자리 선택
	PORTG = (1 << digit);     // 자리 선택
	PORTC = seg[fnd_data[digit]];  // 해당 숫자 출력

	digit++;
	if (digit > 3) digit = 0; // 다음 자리로
}

// ===========================================================
// Timer1 인터럽트 — 1초마다 호출 → 시간 증가용
// ===========================================================
ISR(TIMER1_COMPA_vect)
{
	sec++;
	if (sec >= 60) {
		sec = 0;
		min++;
		if (min >= 60) min = 0;
	}

	// 표시할 값 갱신
	fnd_data[0] = sec % 10;      // 초 1의 자리
	fnd_data[1] = sec / 10;      // 초 10의 자리
	fnd_data[2] = min % 10;      // 분 1의 자리
	fnd_data[3] = min / 10;      // 분 10의 자리
}

// ===========================================================
// Timer 초기화 함수
// ===========================================================
void Timer0_Init(void)
{
	TCCR0 = (1 << CS02) | (1 << CS00);  // 분주비 1024
	TIMSK |= (1 << TOIE0);              // Timer0 Overflow Interrupt Enable
	TCNT0 = 256 - 125;                  // 초기 카운터 설정
}

void Timer1_Init(void)
{
	// CTC 모드 (1초마다 인터럽트 발생)
	TCCR1B = (1 << WGM12) | (1 << CS12) | (1 << CS10); // 분주비 1024
	OCR1A = 15625 - 1;   // (16MHz / 1024 = 15625Hz → 1초마다 인터럽트)
	TIMSK |= (1 << OCIE1A); // Timer1 Compare Match Interrupt Enable
}

// ===========================================================
// 메인 함수
// ===========================================================
int main(void)
{
	DDRC = 0xFF;  // 세그먼트 데이터 (a~g)
	DDRG = 0x0F;  // 자리 선택 출력 (PG0~PG3)

	// 초기 표시값
	fnd_data[0] = 0;
	fnd_data[1] = 0;
	fnd_data[2] = 0;
	fnd_data[3] = 0;

	Timer0_Init(); // 동적 구동용 타이머
	Timer1_Init(); // 1초 시간 증가용 타이머
	sei();         // 전역 인터럽트 허용

	while (1) {
		// 모든 표시와 증가 처리는 인터럽트에서 수행
	}
}
```
